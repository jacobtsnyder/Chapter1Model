```python
import seaborn as sns
import matplotlib.pyplot as plt

df = MacroalgaeGrowth_BrowserHerbivory.pivot(index="rM", columns="hB", values="Cp")

# Format axis labels to 3 decimal places
df.index = [f"{v:.3f}" for v in df.index]
df.columns = [f"{v:.3f}" for v in df.columns]

ax = sns.heatmap(df)
ax.invert_yaxis()  # flip so rC increases from bottom to top

ax.set_title("Coral Cover at 5 years")
ax.set_xlabel("hB")
ax.set_ylabel("μC")

plt.savefig("macro_growth_vs_browser_herbivory_5 years.png", dpi=300, bbox_inches="tight")
plt.show()
```
